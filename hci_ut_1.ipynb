# 1. Introduction to HCI: What is it?
*   **Definition:** HCI involves the design, implementation, and evaluation of interactive systems in the context of the user's task and work.
*   **Core Components:** It focuses on the communication between the **user** (whoever is trying to get the job done), the **computer** (any technology ranging from desktop to embedded systems), and the **interaction** (any direct or indirect communication between the two).
*   **The Three "Use" Words:** For an interactive product to be successful, it must be **useful** (accomplishes what is required), **usable** (does it easily and naturally without danger of error), and **used** (makes people want to use it through attractiveness and engagement).
*   **Multidisciplinary Nature:** The ideal designer needs expertise in psychology and cognitive science (to understand user limitations), ergonomics (for physical capabilities), sociology (for wider context), computer science (to build it), business, and graphic design.

##  The Computer
The computer system comprises various input/output and memory elements that define the boundaries of the interaction.

*   **Text Entry Devices:** Includes the standard alphanumeric keyboard (QWERTY), alternative 26-key layouts (DVORAK), chord keyboards, phone keypads, and even handwriting or speech recognition,.
*   **Pointing and Positioning Devices:** Devices like the mouse, trackball, touchpad, touchscreen, and stylus allow users to point, position, and select items directly on the screen,,.
*   **Display Devices:** Predominantly use bitmap display technologies (like CRT and LCD/flat-panel screens),. Systems are moving beyond the desktop to include large situated displays and digital paper,.
*   **Memory:** Analogous to human memory, computers have short-term memory (RAM) and long-term memory (magnetic and optical disks). Capacity limitations and access methods (how the database requires the user to search) heavily constrain the interface design,.

###**Interaction Styles**
The dialog between user and system can take many forms:
*   **Command line:** Direct textual instructions.
*   **Menus:** Rely on recognition rather than recall by showing options.
*   **Natural language:** Attractive but difficult for machines to interpret due to ambiguity.
*   **Form-fills and spreadsheets:** Great for data entry.
*   **WIMP interface:** The default style for most desktops today, standing for **W**indows, **I**cons, **M**enus, and **P**ointers.



##**Human Factors (Ergonomics)**

The study of the physical characteristics of the interaction. It addresses:
*   **Arrangement of controls:** Logical grouping of controls by function, sequence of use, or frequency of use,,.
*   **Physical environment:** Issues like seating, user size, room to move, and visual angles,.
*   **Health issues:** Designing to minimize physical stress, repetitive strain, temperature extremes, excessive noise, and poor lighting,,.
*   **Use of color:** Color should be distinct, follow cultural conventions (e.g., red for stop/danger), and should not be the only cue due to color-blindness in users,.

##**Cognitive Psychology**

In Human-Computer Interaction, cognitive psychology focuses on understanding the capabilities and limitations of the user's information processing system. A foundational concept is the **Model Human Processor**, which simplifies human interaction into three interacting subsystems: the perceptual system, the motor system, and the cognitive system, each with its own memory and processing capabilities.

**1. Input-Output Channels (Senses and Motor Control)**
*   **Vision:** The primary source of information. Divided into physical reception of the stimulus and the processing/interpretation of that image.
*   **Hearing:** Often considered secondary, but provides vital spatial and environmental information. Sounds can be used to convey attention, status information, confirmation, and navigation.
*   **Touch (Haptic perception):** Important for feedback (e.g., feeling a button depress). Receptors in the skin include thermoreceptors (heat/cold), nociceptors (pain/pressure), and mechanoreceptors (pressure).
*   **Movement:** Involves reaction time and movement time. Speed and accuracy of movement to a target on a screen are modeled by **Fitts' Law**, which states that the time taken to hit a target is a function of the size of the target and the distance that has to be moved.

Fitts' Law is a principle that models the speed and accuracy of movement, specifically the time it takes for a user to move to a target on a screen, such as an icon, menu item, or button
. The law states that the time required to successfully hit a target is a function of two main factors: the size of the target and the distance that must be moved to reach it
.
Mathematically, it is commonly expressed with the formula: Movement time = a + b log2(distance/size + 1)

**2. Human Memory**
*   **Sensory Memory:** Acts as buffers for stimuli received through the senses (e.g., iconic memory for vision, echoic memory for sound),.
*   **Short-term (Working) Memory:** Has a highly limited capacity. Humans can generally remember **7 ± 2 chunks** of information. Successful formation of a chunk is known as closure.
*   **Long-term Memory:** Stores factual knowledge, experiential knowledge, and procedural knowledge. Information is stored associatively in structures like **semantic networks** (classes with inherited attributes), **frames** (data structures for objects), and **scripts** (representing default or stereotypical sequences of events),.

**3. Thinking: Reasoning and Problem Solving**
*   **Reasoning**: Humans process information using three main types of reasoning: **deductive** (drawing logical conclusions from given premises), **inductive** (generalizing from specific cases), and **abductive** (inferring an explanation or cause from an observed fact).
*   **Problem Solving**: Finding solutions to unfamiliar tasks. **Gestalt theory** argues that problem solving involves insight and productive restructuring of the problem. **Problem space theory** argues that users navigate from an initial state to a goal state by selecting operators (actions) and using heuristics like means-ends analysis. People also heavily use **analogy** to map knowledge from a known domain to a new one.
*   **Skill Acquisition**: Humans transition from being novices to experts through practice. According to the **ACT* model**, users move from relying on general-purpose rules (requiring heavy memory access) to developing specific, tuned rules (proceduralization and generalization). Experts group information into larger chunks than novices.

**4. Errors and Mental Models**
Users build theories to understand how systems work, known as **mental models**. These models are often partial, unstable, internally inconsistent, or unscientific. Errors generally fall into two categories:
*   **Slips**: The user formulates the correct action but fails to execute it correctly (e.g., hitting the wrong key).
*   **Mistakes**: The user formulates the wrong goal because they misunderstand the system.

**5. Emotion**
Emotion affects cognitive performance. A biological response to a stimulus (affect) alters how users handle situations. **Positive affect** can stimulate creative thinking and make difficult tasks easier, while **negative affect** pushes users into narrow, focused thinking, making even simple tasks harder.

# **User Centered Design**

The core philosophy of user-centered design can be summarized by a simple mandate: **put the user first, keep the user in the center, and remember the user at the end**.

### Norman’s Seven Principles
Donald Norman summarized the user-centered design philosophy by proposing seven principles for transforming difficult tasks into simple ones:
1. **Use both knowledge in the world and knowledge in the head:** Systems should provide necessary knowledge directly within the environment to assist users, while also allowing experts to internalize routine tasks to increase their efficiency.
2. **Simplify the structure of tasks:** Tasks must be kept simple to avoid excessive memory load and complex problem-solving.
3. **Make things visible:** The interface should clearly show what the system can do, how to achieve it, and the effects of user actions. This bridges the "gulfs of execution and evaluation".
4. **Get the mappings right:** Ensure that user intentions map naturally onto system controls, and that physical controls reflect the magnitude and direction of the task.
5. **Exploit the power of constraints:** Use natural and artificial constraints (like physical shapes or software limitations) to guide the user to make the correct action.
6. **Design for error:** Assume that humans will make errors. Systems should anticipate these errors and feature built-in recovery methods.
7. **When all else fails, standardize:** If there are no natural mappings available, use arbitrary but standardized mappings so users only have to learn them once.

### Shneiderman’s Eight Golden Rules
Ben Shneiderman provides a practical checklist of user-centered design principles that help govern the dialog between user and computer:
1. **Strive for consistency** in terminology, layout, and action sequences.
2. **Enable frequent users to use shortcuts**, such as macros, abbreviations, or special key sequences.
3. **Offer informative feedback** for every action the user takes.
4. **Design dialogs to yield closure**, ensuring the user knows exactly when a task is completed.
5. **Offer error prevention and simple error handling** so users are protected from making mistakes and guided clearly on how to recover if they do.
6. **Permit easy reversal of actions** to relieve user anxiety and encourage them to explore the system.
7. **Support an internal locus of control** so that the user feels they are in charge of the system, rather than the system dictating the interaction.
8. **Reduce short-term memory load** by keeping displays minimalist and avoiding the need to remember information from one screen to the next.

### Nielsen’s Ten Heuristics
Jakob Nielsen’s heuristics are widely used to structure both the design and evaluation of user interfaces:
1. **Visibility of system status:** Keep users informed about what is happening within a reasonable time.
2. **Match between system and the real world:** Speak the user's language rather than using system-oriented terms.
3. **User control and freedom:** Provide clearly marked "emergency exits" for users who choose functions by mistake.
4. **Consistency and standards:** Follow platform conventions so users don't have to wonder if different words or actions mean the same thing.
5. **Error prevention:** A careful design that prevents a problem from occurring is always better than a good error message.
6. **Recognition rather than recall:** Make objects, options, and actions visible so the user doesn't have to rely on their memory.
7. **Flexibility and efficiency of use:** Include unseen accelerators that allow experts to speed up their interactions while keeping the system simple for novices.
8. **Aesthetic and minimalist design:** Remove irrelevant or rarely needed information that competes for the user's attention.
9. **Help users recognize, diagnose, and recover from errors:** Use plain language to indicate problems and constructively suggest solutions.
10. **Help and documentation:** If instructions are necessary, ensure they are focused on the user's task, easy to search, and concisely list concrete steps.

### General Usability Categories
Abstract user-centered design rules and principles generally fall into three overarching categories that define a successful interactive system:
* **Learnability:** Features that allow novice users to quickly understand how to interact with the system and eventually reach maximal performance. This relies on *predictability*, *familiarity* (guessability), *consistency*, and *synthesizability* (assessing the effect of past operations).
* **Flexibility:** The multiplicity of ways users can exchange information with the system. This includes *dialog initiative* (allowing the user freedom from system-imposed constraints), *multi-threading* (supporting multiple tasks at once), and *customizability*.
* **Robustness:** The level of support provided to users to help them achieve and assess their goals. This involves *observability* (evaluating the system's internal state from its representation), *recoverability* from errors, and *responsiveness*.

Evaluation in Human-Computer Interaction is a critical process that occurs throughout the design life cycle, rather than being treated as a final, tacked-on phase. Evaluation has three primary goals: to assess the system's functionality and its accessibility, to understand the user's experience of the interaction, and to identify specific problems or trouble-spots within the design.

The methods used to achieve these goals generally fall into two broad categories: expert analysis and user participation.

### 1. Evaluation Through Expert Analysis
Expert analysis is typically used in the earlier stages of development (like specifications or paper mock-ups) because it is relatively cheap and allows designers to identify and fix expensive mistakes before resources are committed to implementation.

*   **Cognitive Walkthrough:** This method focuses on evaluating how easy a system is to learn through exploration. Evaluators step through a specific sequence of actions required to complete a task and answer four questions for each step: whether the action matches the user's goal, whether the action is visible, whether the user will recognize it as the correct action, and whether the system provides understandable feedback.
*   **Heuristic Evaluation:** Developed by Jakob Nielsen and Rolf Molich, this "discount usability" technique involves multiple experts independently critiquing a system against a set of established guidelines, or heuristics. Evaluators identify potential problems and rate their severity based on principles such as visibility of system status, error prevention, and user control.
*   **Model-Based Evaluation:** This approach integrates evaluation directly into the design framework using cognitive and design models. Techniques like the GOMS (Goals, Operators, Methods, and Selection) model or the keystroke-level model allow designers to predict user performance times and filter out poor design options without needing physical users.
*   **Reviewing Previous Studies:** Evaluators can also review the wealth of existing empirical evidence in psychology and HCI literature to support or refute design decisions, provided the context of the prior experiments matches the current design.

### 2. Evaluation Through User Participation
While expert analysis is valuable, it is not a substitute for actual usability testing with the intended user base. User participation usually occurs later in the development cycle when a working prototype or full implementation is available.

**Styles of Evaluation**
User testing can be conducted in **laboratory studies**, which provide a controlled, interruption-free environment equipped with sophisticated recording tools, though they suffer from an unnatural lack of context. Alternatively, **field studies** observe the user in their natural work environment, capturing real-world interactions and interruptions, though they are subject to noise and are difficult to control.

**Techniques with Users**
*   **Experimental (Empirical) Methods:** This highly controlled method provides empirical evidence to support a specific hypothesis. Evaluators manipulate **independent variables** (such as interface style or number of menu items) to measure the effect on **dependent variables** (such as time taken to complete a task or number of errors). Results are then analyzed using statistical measures.
*   **Observational Methods:** The evaluator watches the user interact with the system. In a **"think aloud"** protocol, the user describes what they are doing and why, providing insight into their decision-making processes. A variation called **cooperative evaluation** encourages the user to act as a collaborator, allowing both the user and evaluator to ask each other questions during the task. These sessions are recorded via paper, audio, video, or computer logging (protocol analysis).
*   **Query Techniques:** The best way to find out how a system meets user requirements is often simply to ask them. **Interviews** provide a flexible, top-down approach to probe deeper into user preferences and attitudes. **Questionnaires** are less flexible but can reach a wider audience quickly, utilizing open-ended, scalar (rating scales), multi-choice, and ranked questions.
*   **Physiological Monitoring:** This emerging area of "objective usability testing" measures bodily responses directly. **Eye tracking** monitors fixations and scan paths to determine how efficiently a user is navigating a visual display. Other physiological measurements include heart rate, blood pressure, and galvanic skin response (sweat), which can indicate levels of stress, arousal, and cognitive effort.

### Choosing an Evaluation Method
Selecting the right evaluation method depends on a careful balancing of several factors:
*   **Stage of the cycle:** Expert analysis suits early design phases, whereas observational and experimental methods require physical prototypes or implementations.
*   **Objectivity vs. Subjectivity:** Methods like cognitive walkthroughs rely heavily on the subjective expertise of the evaluator, whereas controlled experiments or physiological monitoring yield objective, repeatable data.
*   **Measure type:** Techniques are chosen based on whether the team needs qualitative data (non-numeric insights about *why* a problem occurs) or quantitative data (numeric measurements like task completion times).
*   **Resources:** Evaluators must weigh the availability of time, money, recording equipment, and access to representative participants.

## **SUS**

The standard System Usability Scale consists of 10 questions (5 positive and 5 negative) scored on a 5-point scale (1 = Strongly Disagree to 5 = Strongly Agree). To find the overall SUS score (which ranges from 0 to 100):
1. **For odd-numbered questions (positive statements):** Subtract 1 from the user's response (Score - 1).
2. **For even-numbered questions (negative statements):** Subtract the user's response from 5 (5 - Score).
3. **Calculate the total:** Add up all the newly converted values from the 10 questions.
4. **Find the final score:** Multiply the total sum by 2.5 to get the overall SUS score.
